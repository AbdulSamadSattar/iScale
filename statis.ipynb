# Statistics in AI and Data Science

Statistics is the branch of mathematics that helps us collect, organize, analyze, interpret, and present data. In AI and data science, statistics is essential because it turns raw data into meaningful insights, supports model building, helps evaluate performance, and guides decision-making.

## Why statistics matters in AI and data science
- Understand data distribution and patterns
- Detect anomalies and outliers
- Measure relationships between variables
- Make predictions using machine learning models
- Validate experiments and model performance
- Support evidence-based decisions

---

## 1) Descriptive Statistics

Descriptive statistics summarizes and describes the main features of a dataset. It helps us understand what the data looks like without making predictions beyond the dataset.

### Common descriptive measures
- Mean: average value
- Median: middle value
- Mode: most frequent value
- Standard deviation: spread of data around the mean
- Variance: squared spread of data
- Range: difference between maximum and minimum
- Percentiles and quartiles: distribution of values

### Example in AI/data science
If a dataset contains customer ages, descriptive statistics can tell us:
- average age
- youngest and oldest age
- how spread out the ages are
- whether the age distribution is skewed

### Use cases
- Exploratory Data Analysis (EDA)
- Feature understanding
- Data cleaning and quality checks
- Reporting summary insights

---

## 2) Inferential Statistics

Inferential statistics uses a sample of data to make predictions, estimates, or conclusions about a larger population. It is crucial in AI/data science when full population data is not available.

### Key concepts
- Sampling: selecting a subset of data
- Population: the full group of interest
- Sample: the subset used for analysis
- Hypothesis testing: checking if an observed result is statistically significant
- Confidence intervals: estimating a range for a population parameter
- Regression and correlation: understanding relationships between variables

### Example in AI/data science
A team trains a model on a sample of user data to infer how the model might perform on the broader population. They may also test whether a new feature significantly improves model accuracy.

### Use cases
- A/B testing
- Model evaluation
- Estimating population trends
- Generalizing results from sample data
- Statistical significance testing

---

## Descriptive vs Inferential Statistics

| Type | Purpose | Example |
|---|---|---|
| Descriptive Statistics | Summarize observed data | Average salary in a company |
| Inferential Statistics | Draw conclusions about a larger population | Predicting national customer behavior from a survey sample |

---

## In AI and Data Science
Statistics is the foundation of many AI workflows:

- Data preprocessing: handling missing values, outliers, and distributions
- Feature engineering: selecting useful variables
- Model training: estimating parameters and learning patterns
- Model evaluation: accuracy, precision, recall, F1 score, confidence intervals
- Experimentation: testing whether improvements are meaningful
- Bias and variance analysis: understanding model generalization

---

## Summary
Statistics can be divided into:
- Descriptive statistics: describes data
- Inferential statistics: draws conclusions beyond the data

Both are essential in AI and data science because they help us understand patterns, validate models, and make reliable decisions from data.

````python
import math
data = [60, 65, 70, 75, 80]
#by formula

# Calculate the sample standard deviation
sample_std_dev = math.sqrt(sum((x - sum(data) / len(data)) ** 2 for x in data) / (len(data) - 1))
print(f"Sample Standard Deviation (using Math): {sample_std_dev}")

# Calculate the population standard deviation
population_std_dev = math.sqrt(sum((x - sum(data) / len(data)) ** 2 for x in data) / len(data))
print(f"Population Standard Deviation (using Math): {population_std_dev}")
````

````python
import math
data = [60, 65, 70, 75, 80]

# Calculate the sample standard deviation
sample_std_dev = math.sqrt(st.tvar(data))  # by default ddof=1
print(f"Sample Standard Deviation (using Math and Scipy): {sample_std_dev}")

# Calculate the population standard deviation
population_std_dev = math.sqrt(st.tvar(data, ddof=0))   
print(f"Population Standard Deviation (using Math and Scipy): {population_std_dev}")
````

## P-Value Test

The **p-value** measures how likely the observed result (or a more extreme result) would be **if the null hypothesis ($H_0$) were true**.

### Decision Rule

* **p-value ≤ α** → Reject $H_0$ → **Statistically significant**
* **p-value > α** → Fail to reject $H_0$ → **Not statistically significant**

Usually:

$$
\alpha = 0.05
$$

### Example

```text
p-value = 0.03
α = 0.05

0.03 < 0.05 → Reject H₀
```

> **Important:** A p-value does **not** tell you the probability that $H_0$ is true. It tells you how compatible the observed data are with $H_0$.





````python
# Create two random datasets
# Suppose there are test scores of two Groups
group_A = np.random.normal(75,10,30)  # Mean=75, Std Dev=10, Sample Size=30
group_B = np.random.normal(80,12,30)  # Mean=80, Std Dev=12, Sample Size=30

# Perform an independent t-test
t_statistic, p_value = st.ttest_ind(group_A, group_B)

# Display the p-value and t-test
print(f"T-statistic: {t_statistic}")
print(f"P-value: {p_value}")

# Interpretation based on the p-value:
alpha = 0.05  # significance level
if p_value <= alpha:
    print("Reject the null hypothesis: There is a significant difference between the two groups.")
else:
    print("Fail to reject the null hypothesis: There is no significant difference between the two groups.")
````

# Z-Test
$$ Z = \frac{\bar{x} - \mu}{\frac{\sigma}{\sqrt{n}}} $$

Explanation: Used to determine whether a sample mean significantly differs from a known population mean when the population standard deviation $\sigma$ is known (or the sample is sufficiently large).

# T-Test
$$ t = \frac{\bar{x} - \mu}{\frac{s}{\sqrt{n}}} $$

Explanation: Used to determine whether a sample mean significantly differs from a population mean when the population standard deviation is unknown and the sample standard deviation $s$ is used.

Symbols
$\bar{x}$ = Sample mean
$\mu$ = Population mean / hypothesized mean
$\sigma$ = Population standard deviation
$s$ = Sample standard deviation
$n$ = Sample size
$\sqrt{n}$ = Square root of sample size


z table: https://www.w3schools.com/statistics/statistics_z_table.php

Imagine a company claims that the average weight of their products is 500 grams. We want to test if this claim is true by collecting a sample of 30 products. The sample mean is 0 grams, and the population standard deviation is 10 grams. We will test z-test to conduct of the difference in the mean is significant or not.

````python
# Define the known parameters
population_mean = 500 # claimed population mean (mu)
sample_mean = 505 # Sample mean (x̄)
population_std_dev = 10 # Population standard deviation (σ)
sample_size = 30 # Sample size (n)

# Calculate the z-score
z_score = (sample_mean - population_mean) / (population_std_dev / np.sqrt(sample_size))
print(f"Z-score: {z_score}")

# Calculate the p-value for a two-tailed test
p_value = 2 * (1 - st.norm.cdf(abs(z_score)))
print(f"P-value: {p_value}")

# Set Significance level
alpha = 0.05

# Compare p-value with significance level
if p_value < alpha:
    print("Reject the null hypothesis: The sample mean is significantly different from the population mean.")
else:
    print("Fail to reject the null hypothesis: The sample mean is not significantly different from the population mean.")
````
