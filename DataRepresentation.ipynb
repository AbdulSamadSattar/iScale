```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# create a figure and axes
fig, axs = plt.subplots(3, 2, figsize=(14,12) )
fig.suptitle('Histograms with KDE of Various Distribtuions', fontsize=16)
```

```python
fig, axs = plt.subplots(3, 2, figsize=(14,12) )
fig.suptitle('Histograms with KDE of Various Distribtuions', fontsize=16)

# Normal Distribution (Gaussian Distribution)
data_normal = np.random.normal(loc=0, scale=1, size=1000)
sns.histplot(data_normal, kde=True, ax=axs[0, 0], color='blue',stat='density',bins=30)
axs[0, 0].set_title('Normal Distribution', fontsize=14)
# axs[0, 0].set_xlabel('Value', fontsize=12)
# axs[0, 0].set_ylabel('Density', fontsize=12)
axs[0, 0].grid(True)

#skewed Distribution 
data_right_skewed = np.random.exponential(scale=1, size=1000) #Right skewed distribution
data_left_skewed = -np.random.exponential(scale=1, size=1000) #Left skewed distribution
sns.histplot(data_right_skewed, kde=True, ax=axs[0, 1], color='red',stat='density',bins=30,label='Right Skewed')
sns.histplot(data_left_skewed, kde=True, ax=axs[0, 1], color='green',stat='density',bins=30,label='Left Skewed')
axs[0, 1].set_title('Skewed Distribution')
# axs[0, 1].set_xlabel('Value', fontsize=12)  
# axs[0, 1].set_ylabel('Density', fontsize=12)

axs[0, 1].legend()
axs[0, 1].grid(True)


# Uniform Distribution
data_uniform = np.random.uniform(low=-3, high=3, size=1000)
sns.histplot(data_uniform, kde=True, ax=axs[1, 0], color='purple',stat='density',bins=30)
axs[1, 0].set_title('Uniform Distribution', fontsize=14)
axs[1, 0].grid(True)

# Bimodal Distribution
data_bimodal = np.concatenate([np.random.normal(loc=-2, scale=1, size=500), np.random.normal(loc=2, scale=1, size=500)])
sns.histplot(data_bimodal, kde=True, ax=axs[1, 1], color='orange',stat='density',bins=30, label='Bimodal')
axs[1, 1].set_title('Bimodal Distribution', fontsize=14)
axs[1, 1].legend()
axs[1, 1].grid(True)

# Multimodal Distribution
data_multimodal = np.concatenate(
    [np.random.normal(loc=-6, scale=1.5, size=500), 
     np.random.normal(loc=0, scale=1, size=500), 
     np.random.normal(loc=6, scale=1.5, size=500)])
sns.histplot(data_multimodal, kde=True, ax=axs[2, 0], color='brown',stat='density',bins=30, label='Multimodal')
axs[2, 0].set_title('Multimodal Distribution', fontsize=14)
axs[2, 0].grid(True)

# Hide empty subplot (2,1)
axs[2, 1].axis('off')

# Adjust layout
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
